Used for pattern-based trials. i.e. when you want the trigger to be when a specific input has been keyed in.

Used in place of long if-else loops.

#### Key features

* match - evaluates one subject 
* case - represents a possible pattern
* case _ -is the default (used like else in an if-else loop) [NB: there is space before the underscore]

```python
language = input ("Language to learn: ")

match language:
    case "Python" | "R":
        print("Data Science")
    case "JavaScript":
        print("Frontend Web Development")
    case "Java":
        print("Web Development")
    case _ :
        print("General problem solving")
```
